```python
import shutil
import os

# پاک کردن پوشه های fear و disgust از train و test
for folder in ['fear', 'disgust', 'Fear', 'Disgust']:
    path_train = os.path.join('train', folder)
    if os.path.exists(path_train):
        shutil.rmtree(path_train)

    path_test = os.path.join('test', folder)
    if os.path.exists(path_test):
        shutil.rmtree(path_test)

print("حالت های اضافی حذف شدند. حالا فقط ۵ حالت داریم!")
```

```python
import tensorflow as tf

# تنظیمات
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
DATASET_DIR = 'train'

# ۱. لود کردن داده ها
train_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=True, validation_split=0.2, subset="training", seed=123
)
val_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=True, validation_split=0.2, subset="validation", seed=123
)

class_names = train_dataset.class_names
print("دسته بندی ها (حالت ها):", class_names)

# بهینه سازی برای سرعت
AUTOTUNE = tf.data.AUTOTUNE
train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.prefetch(buffer_size=AUTOTUNE)

# ۲. ساخت مدل (استفاده از MobileNetV2)
base_model = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = False

model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(len(class_names), activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# ۳. شروع آموزش!
print("شروع آموزش... لطفا صبر کن...")
history = model.fit(train_dataset, validation_data=val_dataset, epochs=5)

# ۴. ذخیره مدل
model.save('mood_model.h5')
print("مدل با موفقیت ذخیره شد و آماده تست است!")
```

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing import image
import json

# ۱. لود کردن مدلِ آموزش دیده
model = tf.keras.models.load_model('mood_model.h5')

# اسم ۵ حالت (به ترتیب حروف الفبا که مدل یاد گرفته)
class_names = ['angry', 'happy', 'neutral', 'sad', 'surprise']

def predict_emotion(img_path):
    # آماده‌سازی عکس (تغییر سایز به ۲۲۴ در ۲۲۴)
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img)
    img_array = tf.expand_dims(img_array, 0) # اضافه کردن یک بعد برای مدل

    # پیش‌بینی
    predictions = model.predict(img_array)
    score = predictions[0]

    # پیدا کردن بالاترین احتمال
    best_index = np.argmax(score)
    best_mood = class_names[best_index]
    confidence = float(score[best_index])

    # خروجی دقیقاً طبق قرارداد مشترک مسابقه (صفحه ۴ PDF)
    result = {
        "mood": best_mood,
        "confidence": round(confidence, 2)
    }
    return json.dumps(result, ensure_ascii=False)

# --- اسم عکسی که آپلود کردی رو اینجا بنویس ---
test_image = "test.jpg" # اگر اسم عکست چیز دیگه ایه، اینجا تغییرش بده

print(f"نتیجه تشخیص برای عکس {test_image}:")
print(predict_emotion(test_image))
```

```python
# ۱. نصب کتابخانه گرادیو
!pip install -q gradio

import gradio as gr
import tensorflow as tf
import numpy as np
import json

# ۲. لود کردن مدل
model = tf.keras.models.load_model('mood_model.h5')
class_names = ['angry', 'happy', 'neutral', 'sad', 'surprise']

# ترجمه حالت ها برای نمایش به کاربر
fa_moods = {
    "angry": "عصبانی 😠",
    "happy": "شاد 😄",
    "neutral": "خنثی 😐",
    "sad": "غمگین 😢",
    "surprise": "متعجب 😲"
}

# ۳. تابعی که با آپلود عکس اجرا میشه
def predict_mood_ui(img):
    if img is None:
        return "لطفا یک عکس آپلود کنید."

    # تغییر سایز عکس
    img = img.resize((224, 224))
    img_array = np.array(img)
    img_array = tf.expand_dims(img_array, 0)

    # پیش بینی
    predictions = model.predict(img_array)
    score = predictions[0]
    best_index = np.argmax(score)
    best_mood = class_names[best_index]
    confidence = float(score[best_index])

    # خروجی خوشگل برای نمایش در صفحه وب
    result_text = f"حال تشخیص داده شده: {fa_moods[best_mood]}\n\nمیزان اطمینان: {round(confidence*100, 1)}%"
    return result_text

# ۴. ساخت رابط کاربری
demo = gr.Interface(
    fn=predict_mood_ui,
    inputs=gr.Image(type="pil", label="عکس چهره را آپلود کنید"),
    outputs=gr.Textbox(label="نتیجه تشخیص سینمود"),
    title="سینمود - دستیار هوشمند تشخیص حال",
    description="عکس چهره خود را آپلود کنید تا هوش مصنوعی حال شما را تشخیص دهد.",
    theme="default"
)

# ۵. اجرای صفحه وب (ایجاد یک لینک عمومی)
demo.launch(share=True)
```
