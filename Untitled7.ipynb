```python
import shutil
import os

# باز کردن فایل زیپ
!unzip -q archive.zip

# پاک کردن حالت های fear و disgust برای هماهنگی با ۵ حال سینمود
for folder in ['fear', 'disgust', 'Fear', 'Disgust']:
    path_train = os.path.join('train', folder)
    if os.path.exists(path_train):
        shutil.rmtree(path_train)
    path_test = os.path.join('test', folder)
    if os.path.exists(path_test):
        shutil.rmtree(path_test)

print("دیتاست آماده شد. فقط ۵ حالت داریم!")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
DATASET_DIR = 'train'

train_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=True, validation_split=0.2, subset="training", seed=123
)
val_dataset = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR, image_size=IMG_SIZE, batch_size=BATCH_SIZE,
    label_mode='categorical', shuffle=True, validation_split=0.2, subset="validation", seed=123
)

class_names = train_dataset.class_names
print("دسته بندی ها:", class_names)

AUTOTUNE = tf.data.AUTOTUNE
train_dataset = train_dataset.prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.prefetch(buffer_size=AUTOTUNE)
```

```python
# ساخت مدل با MobileNetV2
base_model = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = False

model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(len(class_names), activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("شروع آموزش... لطفا صبر کنید...")
history = model.fit(train_dataset, validation_data=val_dataset, epochs=5)

# ذخیره مدل
model.save('mood_model.h5')
print("مدل با موفقیت ذخیره شد!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

print("در حال محاسبه ماتریس خطا...")
y_true = []
y_pred = []

for images, labels in val_dataset:
    predictions = model.predict(images, verbose=0)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(predictions, axis=1))

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix - Cinemood AI')
plt.xlabel('Predicted Mood')
plt.ylabel('True Mood')
plt.show()

print("\n--- Classification Report ---\n")
print(classification_report(y_true, y_pred, target_names=class_names))
```

### مقایسه معماری‌ها

در این پروژه از مدل **MobileNetV2** استفاده شد. در مقایسه با معماری‌های سنگین‌تر مثل ResNet50 یا VGG16، مدل MobileNetV2 حجم بسیار کمتری دارد و سرعت اجرای بالاتری روی CPU و GPU ارائه می‌دهد.

از آنجا که هدف این پروژه اجرای زنده (Real-time) در مرورگر است، این سرعت حیاتی بود. اگرچه دقت نهایی ممکن است کمی کمتر از مدل‌های سنگین باشد، اما سرعت و سبکی آن برای یک اپلیکیشن وب یک مزیت بسیار بزرگ است.

```python
import json
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing import image

# لود کردن مدل تشخیص چهره OpenCV
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

def test_single_photo(img_path):
    try:
        # خواندن عکس با OpenCV
        img_cv = cv2.imread(img_path)
        if img_cv is None:
            print(f"عکس {img_path} پیدا نشد.")
            return

        gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.1, 4)

        # اگر چهره پیدا شد، برشش می‌دهیم
        if len(faces) > 0:
            (x, y, w, h) = faces[0]
            face_roi = img_cv[y:y+h, x:x+w]
            face_resized = cv2.resize(face_roi, (224, 224))
            face_rgb = cv2.cvtColor(face_resized, cv2.COLOR_BGR2RGB)
        else:
            # اگر چهره پیدا نشد، کل عکس را تغییر سایز می‌دهیم
            face_rgb = cv2.cvtColor(cv2.resize(img_cv, (224, 224)), cv2.COLOR_BGR2RGB)

        img_array = tf.expand_dims(np.array(face_rgb, dtype=np.float32), 0)

        # پیش‌بینی
        predictions = model.predict(img_array, verbose=0)
        score = predictions[0]
        best_index = np.argmax(score)
        best_mood = class_names[best_index]
        confidence = float(score[best_index])

        # ساخت خروجی JSON
        result = {"mood": best_mood, "confidence": round(confidence, 2)}
        print(f"عکس {img_path} -> خروجی: {json.dumps(result)}")

    except Exception as e:
        print(f"خطا در {img_path}: {e}")

print("--- شروع تست روی ۳ عکس ---")
test_single_photo("photo1.webp")
test_single_photo("photo2.webp")
test_single_photo("photo3.webp")
print("---------------------------")
```

```python
# ۱. نصب و راه‌اندازی OpenCV
!pip install -q opencv-python-headless
import cv2
import numpy as np
import gradio as gr
import tensorflow as tf

fa_moods = {
    "angry": "عصبانی 😠", "happy": "شاد 😄", "neutral": "خنثی 😐",
    "sad": "غمگین 😢", "surprise": "متعجب 😲"
}

# ۲. لود کردن مدل تشخیص چهره از OpenCV
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

def predict_mood_ui(img):
    if img is None:
        return "لطفا یک عکس آپلود کنید یا با وبکم عکس بگیرید."

    # تبدیل عکس برای OpenCV
    img_cv = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)
    gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)

    # پیدا کردن چهره در عکس
    faces = face_cascade.detectMultiScale(gray, 1.1, 4)

    if len(faces) > 0:
        # برش دادن کادر چهره
        (x, y, w, h) = faces[0]
        face_roi = img_cv[y:y+h, x:x+w]
        face_resized = cv2.resize(face_roi, (224, 224))
        face_rgb = cv2.cvtColor(face_resized, cv2.COLOR_BGR2RGB)
        img_array = tf.expand_dims(np.array(face_rgb, dtype=np.float32), 0)
    else:
        # اگر چهره پیدا نشد، کل عکس رو بفرست
        img_resized = img.resize((224, 224))
        img_array = tf.expand_dims(np.array(img_resized, dtype=np.float32), 0)

    # ۳. پیش‌بینی با مدل
    predictions = model.predict(img_array)
    score = predictions[0]
    best_index = np.argmax(score)
    best_mood = class_names[best_index]
    confidence = float(score[best_index])

    return f"حال تشخیص داده شده: {fa_moods[best_mood]}\n\nمیزان اطمینان: {round(confidence*100, 1)}%"

# ۴. ساخت رابط کاربری
demo = gr.Interface(
    fn=predict_mood_ui,
    inputs=gr.Image(type="pil", sources=["upload", "webcam"], label="عکس آپلود کنید یا با وبکم عکس بگیرید"),
    outputs=gr.Textbox(label="نتیجه تشخیص سینمود"),
    title="سینمود - دستیار هوشمند تشخیص حال (با OpenCV)",
    description="عکس چهره خود را آپلود کنید یا از وبکم استفاده کنید تا هوش مصنوعی حال شما را تشخیص دهد."
)

demo.launch(share=True)
```
